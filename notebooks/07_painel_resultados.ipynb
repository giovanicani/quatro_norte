```python
import pandas as pd
from pathlib import Path
from IPython.display import Image, display
PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "notebooks": PROJECT_ROOT = PROJECT_ROOT.parent
TABLES = PROJECT_ROOT/"reports"/"tables"; FIG = PROJECT_ROOT/"reports"/"figures"
def show_tab(n):
    p = TABLES/n
    print("==", n, "=="); print(pd.read_csv(p).to_string(index=False) if p.exists() else "(ausente)")
def show_fig(rel):
    p = FIG/rel
    display(Image(str(p))) if p.exists() else print("(figura ausente)", rel)
```
